```python
import numpy as np  # Importe la bibliothèque numpy pour les calculs matriciels

class RNN:  # Définit une classe pour un Réseau de Neurones Récurrent
    def __init__(self, input_dim, hidden_dim, output_dim):
        # Initialise les poids du RNN avec des valeurs aléatoires
        self.Wxh = np.random.randn(input_dim, hidden_dim)  # Poids entrée → couche cachée
        self.Whh = np.random.randn(hidden_dim, hidden_dim)  # Poids récursif (couche cachée)
        self.Why = np.random.randn(hidden_dim, output_dim)  # Poids couche cachée → sortie
        self.h = np.zeros((1, hidden_dim))  # État caché initial (1 batch, hidden_dim neurones)

    def forward(self, x):
        # Transforme l'entrée x (index) en vecteur one-hot
        x_onehot = np.zeros((1, len(self.Wxh)))  # Vecteur de taille input_dim
        x_onehot[0, x] = 1  # Met à 1 l'index correspondant à x

        # Calcul de l'état caché suivant
        self.h = np.tanh(x_onehot @ self.Wxh + self.h @ self.Whh)  # Activation tanh
        y = self.h @ self.Why  # Calcul de la sortie (logits)
        return y

    def predict(self, sequence):
        # Prédiction sur une séquence d'entrées
        outputs = []  # Liste pour stocker les sorties
        for x in sequence:
            output = self.forward(x)  # Applique forward à chaque élément
            outputs.append(output)
        return outputs

# Exemple
vocab_size = 5  # Taille du vocabulaire
hidden_dim = 2   # Nombre de neurones cachés
rnn = RNN(vocab_size, hidden_dim, vocab_size)  # Crée un RNN

# Test
sequence = [1, 2]  # Séquence d'entrée ("hello")
outputs = rnn.predict(sequence)
print(outputs)  # Affiche les logits (à softmax pour des probas)
```
